### 2. Wrangle individual datasets

#### ✅ 2.1 Update data types
* Convert the `Date` column in each dataframe so that values are datetime objects. 
* List the date range for the two Ozone dataframes.

````python
# Convert the Date column to DateTime objects
NC2024['Date'].head
NC2024['Date']= pd.to_datetime(NC2024['Date'], format='%m/%d/%Y')
NC2025['Date']= pd.to_datetime(NC2025['Date'], format='%m/%d/%Y')
PM25_NC2024['Date']= pd.to_datetime(PM25_NC2024['Date'], format='%m/%d/%Y')
PM25_NC2024['Date']= pd.to_datetime(PM25_NC2025['Date'], format='%m/%d/%Y')
# Print date ranges
print("NC2024 Date range", NC2024['Date'].min(), "to", NC2024['Date'].max())
print("NC2025 Date range", NC2025['Date'].min(), "to", NC2025['Date'].max())
````

#### ✅ 2.2 Eliminate unnecessary columns
* Subset each dataframe to include only the following columns:  
```'Date', 'Daily AQI Value', 'Local Site Name', 'AQS Parameter Description', 'County', 'Site Latitude', 'Site Longitude'```

````python
# 2.2 Subset columns
col_to_keep = ['Date', 'Daily AQI Value', 'Local Site Name', 'AQS Parameter Description', 'County', 'Site Latitude', 'Site Longitude']
NC2024=NC2024.loc[:, col_to_keep]
NC2025=NC2025.loc[:, col_to_keep]
PM25_NC2024=PM25_NC2024.loc[:, col_to_keep]
PM25_NC2025=PM25_NC2025.loc[:, col_to_keep]
````

* For the PM2.5 datasets, fill all cells in the `AQS Parameter Description` with the string "PM2.5".  
(*All cell values in this column will be identical.*)

````python
# 2.3 Update values
PM25_NC2024.loc[:, 'AQS Parameter Description']="PM2.5"
PM25_NC2025.loc[:, 'AQS Parameter Description']="PM2.5"
````

### 4. Wrangle the combined dataset

#### ✅ 4.1 Subset rows 
* Generate a list of local site names common to all 4 dataframes (code provided, but replace the dataframe names to match your own).
* Create a subset of the full dataset, selecting those that match these local site names.
* Report the dimensions of the resulting dataframe. (This should be `18,778 x 7`)

````python
# Generate a list of local site names common to all 4 dataframes
## Create set objects from the values in the Local Site Name columns
sites_oz24 = set(NC2024['Local Site Name']) #<-- update dataframe name to use your 2024 Ozone dataframe
sites_oz25 = set(NC2025['Local Site Name'])
sites_pm24 = set(PM25_NC2024['Local Site Name'])
sites_pm25 = set(PM25_NC2025['Local Site Name'])
## Compute the intersection 
common_sites = sites_oz24 & sites_oz25 & sites_pm24 & sites_pm25
common_sites
````

````python
# 4.1 Subset for records with sites in common
EPA_air_common_sites=EPA_air_combined[EPA_air_combined['Local Site Name'].isin(common_sites)]
EPA_air_common_sites.shape
````

#### ✅ 4.2 Group and summarize data 
We want a dataset that lists one record per measured pollutant per location per day. To ensure we have that, we'll group data by date, local site name, AQS parameter, and county, computing the mean daily AQI value as well as mean latitude and longitude value. 
* Group the data on the grouping variables mentioned above. 
* For each group, compute mean daily AQI value, mean latitude, and mean longitude. 
* Reset the index of the summarized dataset to restore the product to a flat dataframe.
* Report the dimensions of the dataset. (This should be `17,017 x 7`.)

````python
# 4.2 Group and summarize data, then report its dimensions
EPA_air_group=(
    EPA_air_common_sites
    .groupby(['Date', 'Local Site Name', 'AQS Parameter Description', 'County'])
    .agg(
        mean_aqi=("Daily AQI Value", "mean"),
        mean_latitude=("Site Latitude", "mean"),
        mean_longitude=("Site Longitude", "mean"),
    )
.reset_index()
)

EPA_air_group.shape
````

#### ✅ 4.3 Add calculated columns
* Add `Year` and `Month` columns to your dataframe, deriving values from the `Date` field. 

````python
# 4.3 Calculate year and month columns
EPA_air_group['Date']=pd.to_datetime(EPA_air_group['Date'])

EPA_air_group=(
    EPA_air_group
    .assign(
        Year= lambda df:df['Date'].dt.year,
        Month= lambda df:df['Date'].dt.month
    )
)
EPA_air_group.head()
````

#### ✅ 4.4 Spread AQI values into separate columns: OZone & PM2.5
* Reshape our table so that Daily AQI values for Ozone and PM2.5 data appear in separate columns.
* Reset the index of the reshaped table to restore it as a flat dataframe.
* Display the first 5 records in the resulting dataframe 

````python
# 4.4 Spread data so Ozone, PM2.5 daily AQI valiues occupy separate columns
EPA_air_wide=( 
    EPA_air_group.pivot_table(
        index=['Date', 'Local Site Name', 'County', 'Year', 'Month', 'mean_latitude', 'mean_longitude'],
        columns='AQS Parameter Description',
        values = 'mean_aqi'
    )
.reset_index()
)
EPA_air_wide.head()
````

## 6. Summarize and explore the data

#### ✅ 6.1 Generate a month summary table
* Group the above table to compute mean Ozone and PM2.5 values for each unique site, month, year combination. Be sure to reset the index of the product. 
* Display the first 5 records of this table.

````python
# 6.1 Generate a summary table and show its first 5 records
EPA_air_month_summary = (
    EPA_air_wide
    .groupby(['Local Site Name', 'Year', 'Month'])
    .agg(
        mean_ozone=('Ozone', 'mean'),
        mean_pm25=('PM2.5', 'mean')
    )
    .reset_index()
)
EPA_air_month_summary.head()
````
